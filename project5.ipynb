```python
import math as m
def aval(x):
    y=m.sin(x)+1
    v=sqrt(y)
    return(v)
aval(30)
```

```python
def dovom(b):
    l=sin(b)/cos(b)
    j=sqrt(l)
    return(j)
dovom(30)
```

```python
def sevom(p):
    m=tan(p)/sin(p)
    n=sqrt(m)
    return(n)
sevom(30)
```

```python
def chahar(e,x):
    d=1/1+e**-x
    o=sqrt(d)
    return(o)
chahar(10,20)
```

```python
def panj(w,a):
    c=sqrt(w)/sqrt(a)
    z=w-a
    x=z*c
    return(x)
panj(50,10)
```

```python
def op(e,x):
    b=1-e**sin(x)
    return(b)
op(10,30)
```
