```python
def jam(s):
    y=s
    return(y)
jam(2)
```

```python
def zarb(x):
    p=x
    return(p)
zarb(1)
```

```python
def nn(a,b,x,c):
    t=a*x**2+b*x+c
    return(t)
nn(2,2,3,4)
```

```python
def jj(a,b,x,c,d):
    m=a*x**3+b*x**2+c*x+d
    return(m)
jj(2,2,3,4,5)
```

```python
def tt(a,x):
    o=a**x
    return(o)
tt(2,2)
```
